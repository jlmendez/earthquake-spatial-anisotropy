```python
rng=np.random.default_rng(42)
n=600
s=rng.normal(0,5,n); cross=rng.normal(0,1.2,n)
strike=np.deg2rad(125)
x=s*np.sin(strike)+cross*np.cos(strike)
y=s*np.cos(strike)-cross*np.sin(strike)
lat0,lon0=14.45,-90.69
lat=lat0+y/111.32
lon=lon0+x/(111.32*np.cos(np.deg2rad(lat0)))
principal_orientation(lat,lon)
```

```python
compare_corridors(lat,lon,[90,110,125,140,160],half_width_km=2.5)
```

```python
labels, model=cluster_epicenters(lat,lon,n_clusters=2)
np.bincount(labels)
```
